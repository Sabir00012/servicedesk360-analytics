```python
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns
df=pd.read_csv('../01_Dataset/tickets.csv', parse_dates=['created_date'])
df.head(); df['status'].value_counts()
```

```python
df[df.status=='Resolved'].groupby('category')['resolution_hours'].mean().sort_values().plot(kind='barh')
plt.title('Avg Resolution Hours by Category'); plt.show()
sns.boxplot(data=df[df.status=='Resolved'], x='priority', y='resolution_hours')
plt.title('Resolution Time by Priority'); plt.show()
```
